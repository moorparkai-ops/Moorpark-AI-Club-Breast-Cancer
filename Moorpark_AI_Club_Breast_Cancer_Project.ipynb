### 📂 Loading Our Shared Dataset

Before exploring the data, we need to load our breast cancer dataset into Google Colab.

We'll use two Python libraries (a reusable collection of pre-written code—including modules, functions, classes, and data values—that allows you to add specific functionality to your programs without writing it from scratch):

- **`gdown`** — Downloads our shared CSV file directly from Google Drive, so everyone can access the same dataset without uploading it manually.
- **`pandas`** — Reads the CSV file and converts it into a DataFrame (a table of rows and columns) that we can analyze.

**Our goal:** Load the dataset and check how many patient records and columns it contains.

```python

import pandas as pd
import gdown

# Our shared Google Drive dataset
file_id = "12wvpizwcMn3Ogey4X_WGBDuNxEVN7LC3"

# Download the CSV into Colab
gdown.download(
    id=file_id,
    output="Breast_Cancer.csv",
    quiet=False
)

# Load the dataset
df = pd.read_csv("Breast_Cancer.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

Step 3: Understanding Our Dataset Structure

Before analyzing the data, we need to understand how it's organized.

We'll use three Pandas commands:

df.shape — Shows the number of rows (patients) and columns (variables).

df.columns — Displays the names of all columns.

df.info() — Shows each column's data type and the number of non-missing values.

Why is this important?

Machine learning algorithms require properly formatted data. Understanding our variables helps us identify which features may need preprocessing before building a model.

```python

# Number of rows and columns
print("Dataset shape:", df.shape)

# Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display information about each column
print("\nDataset Information:")
df.info()
```

### Step 4: Checking for Missing Values

Real-world datasets often contain missing information. For example, a patient's tumor size or cancer stage might not have been recorded.

Before training a machine learning model, we need to identify these missing values.

We'll use two Pandas commands:

- **`df.isnull()`** — Checks whether each value is missing.
- **`df.isnull().sum()`** — Counts the missing values in each column.

**Why is this important?**

Missing values can affect machine learning models. Depending on the situation, we may need to fill in missing values (imputation) or remove certain records.

```python

# Check missing values in each column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

# Total number of missing values
print("\nTotal missing values:",
      missing_values.sum())
```
